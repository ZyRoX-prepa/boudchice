## 1. Importation des Modules Nécessaires


```python
import os
import numpy as np
import pandas as pd
import scipy.stats as sps
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.linear_model import LinearRegression

# Configuration for better visualizations
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 11
```

## 2. Chargement et Lecture des Données

### Option 1: Créer des données d'exemple
Si vous n'avez pas le fichier CSV, vous pouvez utiliser les données d'exemple ci-dessous.


```python
# Option 1: Créer des données d'exemple pour la démonstration
# Ces données représentent l'Espérance de vie vs Taux de Fertilité par pays

np.random.seed(42)
n_samples = 50

# Générer des données avec corrélation négative
esp_vie = np.random.normal(70, 10, n_samples)
taux_fert = 5 - 0.04 * esp_vie + np.random.normal(0, 2, n_samples)

# Créer le DataFrame
D1 = pd.DataFrame({
    'An': np.arange(1980, 1980 + n_samples),
    'esp vie': esp_vie,
    'taux fert': taux_fert
})

# Afficher les premières lignes
print("Premières lignes du dataset:")
print(D1.head(10))
print(f"\nShape du dataset: {D1.shape}")
```

## 3. Préparation des Données


```python
# Sélectionner les colonnes intéressantes
table1 = D1[['An', 'taux fert', 'esp vie']]

# Renommer les colonnes pour plus de clarté
table1 = table1.rename(columns={
    'taux fert': 'Taux de Fertilité',
    'esp vie': 'Espérance de vie'
})

# Supprimer les lignes avec des données manquantes
table1 = table1.dropna()

print("Données préparées:")
print(table1.head())
print(f"\nShape: {table1.shape}")
```

## Question 1: Centralité et Dispersion

**Renvoyer à l'aide des commandes appropriées:**
- Le mode
- La médiane
- La moyenne empirique
- La boîte à moustaches (boxplot)

Pour les deux variables: **'Taux de Fertilité'** et **'Espérance de vie'**


### Statistiques pour 'Taux de Fertilité'


```python
# Extraction de la variable
taux_fert = table1['Taux de Fertilité']

# Statistiques de centralité et dispersion
print("=== STATISTIQUES POUR 'TAUX DE FERTILITÉ' ===")
print(f"Nombre de valeurs: {taux_fert.count()}")
print(f"Moyenne: {taux_fert.mean():.4f}")
print(f"Médiane: {taux_fert.median():.4f}")
print(f"Mode: {taux_fert.mode().values[0]:.4f}" if len(taux_fert.mode()) > 0 else "Mode: Pas de mode unique")
print(f"Écart-type: {taux_fert.std():.4f}")
print(f"Variance: {taux_fert.var():.4f}")
print(f"Minimum: {taux_fert.min():.4f}")
print(f"Maximum: {taux_fert.max():.4f}")
print(f"Q1 (25%): {taux_fert.quantile(0.25):.4f}")
print(f"Q3 (75%): {taux_fert.quantile(0.75):.4f}")
print(f"IQR (Écart interquartile): {taux_fert.quantile(0.75) - taux_fert.quantile(0.25):.4f}")

# Statistiques descriptives complètes
print("\nRésumé statistique complet:")
print(taux_fert.describe())
```

### Statistiques pour 'Espérance de vie'


```python
# Extraction de la variable
esp_vie = table1['Espérance de vie']

# Statistiques de centralité et dispersion
print("=== STATISTIQUES POUR 'ESPÉRANCE DE VIE' ===")
print(f"Nombre de valeurs: {esp_vie.count()}")
print(f"Moyenne: {esp_vie.mean():.4f}")
print(f"Médiane: {esp_vie.median():.4f}")
print(f"Mode: {esp_vie.mode().values[0]:.4f}" if len(esp_vie.mode()) > 0 else "Mode: Pas de mode unique")
print(f"Écart-type: {esp_vie.std():.4f}")
print(f"Variance: {esp_vie.var():.4f}")
print(f"Minimum: {esp_vie.min():.4f}")
print(f"Maximum: {esp_vie.max():.4f}")
print(f"Q1 (25%): {esp_vie.quantile(0.25):.4f}")
print(f"Q3 (75%): {esp_vie.quantile(0.75):.4f}")
print(f"IQR (Écart interquartile): {esp_vie.quantile(0.75) - esp_vie.quantile(0.25):.4f}")

# Statistiques descriptives complètes
print("\nRésumé statistique complet:")
print(esp_vie.describe())
```

### Boîtes à Moustaches (Boxplots)


```python
# Créer les boxplots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Boxplot pour Taux de Fertilité
table1.boxplot(column='Taux de Fertilité', ax=axes[0])
axes[0].set_title('Boîte à Moustaches - Taux de Fertilité', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Taux de Fertilité')
axes[0].grid(True, alpha=0.3)

# Boxplot pour Espérance de vie
table1.boxplot(column='Espérance de vie', ax=axes[1])
axes[1].set_title('Boîte à Moustaches - Espérance de vie', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Espérance de vie')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Question 2: Affichage des Données et Point Moyen

**Comment afficher également le point moyen du nuage (X, Y)?**


```python
# Extraction des variables
X = table1['Espérance de vie']
Y = table1['Taux de Fertilité']

# Calculer le point moyen
X_mean = X.mean()
Y_mean = Y.mean()

print(f"Point moyen: ({X_mean:.2f}, {Y_mean:.2f})")

# Créer le nuage de points avec le point moyen
plt.figure(figsize=(10, 7))

# Nuage de points
plt.scatter(X, Y, marker='+', s=100, alpha=0.6, label='Points observés')

# Point moyen
plt.scatter(X_mean, Y_mean, marker='o', s=300, color='red', 
            edgecolors='darkred', linewidth=2, label='Point moyen', zorder=5)

plt.grid(True, alpha=0.3)
plt.xlabel('Espérance de vie (X)', fontsize=12)
plt.ylabel('Taux de Fertilité (Y)', fontsize=12)
plt.title('Nuage de Points avec Point Moyen', fontsize=14, fontweight='bold')
plt.legend(fontsize=11)
plt.show()
```

## Question 3: Corrélation Linéaire

**À l'aide des commandes Python:**
- Calculer la covariance de X et Y notée cov(x,y)
- Calculer le coefficient de corrélation r
- Interpréter la relation entre X et Y selon la valeur de r


```python
# Calcul de la covariance
covariance = np.cov(X, Y)[0, 1]
print(f"Covariance cov(X, Y) = {covariance:.4f}")

# Calcul du coefficient de corrélation de Pearson
correlation_pearson = X.corr(Y)
print(f"\nCorrélation de Pearson r = {correlation_pearson:.4f}")

# Utiliser scipy pour obtenir p-value
corr_coefficient, p_value = stats.pearsonr(X, Y)
print(f"p-value = {p_value:.6f}")

# Interprétation
print("\n=== INTERPRÉTATION ===")
if abs(correlation_pearson) > 0.8:
    force = "très forte"
elif abs(correlation_pearson) > 0.6:
    force = "forte"
elif abs(correlation_pearson) > 0.4:
    force = "modérée"
elif abs(correlation_pearson) > 0.2:
    force = "faible"
else:
    force = "très faible"

direction = "positive" if correlation_pearson > 0 else "négative"
significance = "significative" if p_value < 0.05 else "non significative"

print(f"La corrélation est {force} et {direction}.")
print(f"Elle est {significance} (p-value = {p_value:.6f})")

if correlation_pearson < 0:
    print(f"\nInterprétation: Quand l'Espérance de vie AUGMENTE, "
          f"le Taux de Fertilité DIMINUE.")
else:
    print(f"\nInterprétation: Quand l'Espérance de vie AUGMENTE, "
          f"le Taux de Fertilité AUGMENTE.")
```

### Matrice de Corrélation


```python
# Calculer la matrice de corrélation
corr_matrix = table1.corr()
print("Matrice de Corrélation:")
print(corr_matrix)

# Visualiser la matrice de corrélation
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1, fmt='.3f')
plt.title('Matrice de Corrélation', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Question 4: Régression Linéaire

**Écrire un code Python permettant de représenter la droite de régression linéaire de Y en X au-dessus du nuage de points (X, Y)**


```python
# Préparer les données pour la régression
X_array = X.values.reshape(-1, 1)  # Reshape pour sklearn
Y_array = Y.values

# Créer et ajuster le modèle de régression linéaire
model = LinearRegression()
model.fit(X_array, Y_array)

# Coefficients
slope = model.coef_[0]
intercept = model.intercept_

print(f"Équation de la droite de régression:")
print(f"Y = {intercept:.4f} + {slope:.4f} * X")
print(f"\nOù:")
print(f"  - Pente (slope) = {slope:.4f}")
print(f"  - Ordonnée à l'origine (intercept) = {intercept:.4f}")

# Calculer R²
R_squared = model.score(X_array, Y_array)
print(f"\nCoefficient de détermination R² = {R_squared:.4f}")
print(f"Cela signifie que {R_squared*100:.2f}% de la variance de Y ")
print(f"est expliquée par X.")

# Prédictions
Y_pred = model.predict(X_array)
```

### Visualisation de la Droite de Régression


```python
# Créer le graphique
plt.figure(figsize=(12, 7))

# Nuage de points
plt.scatter(X, Y, marker='+', s=100, alpha=0.6, label='Points observés')

# Droite de régression
X_line = np.array([X.min(), X.max()]).reshape(-1, 1)
Y_line = model.predict(X_line)
plt.plot(X_line, Y_line, 'r-', linewidth=2.5, label='Droite de régression')

# Point moyen
plt.scatter(X_mean, Y_mean, marker='o', s=300, color='green', 
            edgecolors='darkgreen', linewidth=2, label='Point moyen', zorder=5)

# Résidus (optionnel - montrer les écarts)
for i in range(len(X)):
    plt.plot([X.iloc[i], X.iloc[i]], [Y.iloc[i], Y_pred[i]], 
             'k--', alpha=0.2, linewidth=0.5)

plt.grid(True, alpha=0.3)
plt.xlabel('Espérance de vie (X)', fontsize=12)
plt.ylabel('Taux de Fertilité (Y)', fontsize=12)
plt.title(f'Régression Linéaire: Y = {intercept:.2f} + {slope:.4f}*X (R² = {R_squared:.4f})',
          fontsize=14, fontweight='bold')
plt.legend(fontsize=11, loc='best')
plt.tight_layout()
plt.show()
```

### Analyse des Résidus


```python
# Calcul des résidus
residuals = Y_array - Y_pred

print("=== ANALYSE DES RÉSIDUS ===")
print(f"Moyenne des résidus: {np.mean(residuals):.6f}")
print(f"Écart-type des résidus: {np.std(residuals):.4f}")
print(f"Minimum des résidus: {np.min(residuals):.4f}")
print(f"Maximum des résidus: {np.max(residuals):.4f}")

# Visualisation des résidus
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Graphique des résidus vs valeurs prédites
axes[0].scatter(Y_pred, residuals, alpha=0.6)
axes[0].axhline(y=0, color='r', linestyle='--', linewidth=2)
axes[0].set_xlabel('Valeurs Prédites (Ŷ)')
axes[0].set_ylabel('Résidus (e)')
axes[0].set_title('Graphique des Résidus vs Prédictions')
axes[0].grid(True, alpha=0.3)

# Histogramme des résidus
axes[1].hist(residuals, bins=15, edgecolor='black', alpha=0.7)
axes[1].set_xlabel('Résidus')
axes[1].set_ylabel('Fréquence')
axes[1].set_title('Distribution des Résidus')
axes[1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
```
